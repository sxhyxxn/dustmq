```python
from qgis.core import (
    QgsVectorLayer, QgsProject, QgsFeature, QgsGeometry
)
import os

# 경로 설정
base_path = r"C:/Users/suhyeon/Desktop/aoq"
line_path = os.path.join(base_path, "line.shp")       # 선 레이어 (파일명에 맞게 수정)
boundary_path = os.path.join(base_path, "boundary.shp")  # 경계 폴리곤 레이어 (파일명에 맞게 수정)

# 레이어 불러오기
line_layer = QgsVectorLayer(line_path, "LineLayer", "ogr")
boundary_layer = QgsVectorLayer(boundary_path, "BoundaryLayer", "ogr")

# 프로젝트에 레이어 추가
QgsProject.instance().addMapLayer(line_layer)
QgsProject.instance().addMapLayer(boundary_layer)

# 자른 선을 담을 새 메모리 레이어
trimmed_layer = QgsVectorLayer("LineString?crs=" + line_layer.crs().authid(), "TrimmedLines", "memory")
prov = trimmed_layer.dataProvider()

# 첫 번째 경계 폴리곤 사용
boundary_feature = next(boundary_layer.getFeatures())
boundary_geom = boundary_feature.geometry()

# 선 잘라서 결과 레이어에 추가
for feat in line_layer.getFeatures():
    geom = feat.geometry()
    trimmed_geom = geom.intersection(boundary_geom)
    if not trimmed_geom.isEmpty():
        new_feat = QgsFeature()
        new_feat.setGeometry(trimmed_geom)
        prov.addFeature(new_feat)

# 결과 레이어 추가
QgsProject.instance().addMapLayer(trimmed_layer)

```
